```python
# Consultamos todas las tablas del esquema SQLite
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
tablas = [fila["name"] for fila in cursor.fetchall()]
print("Tablas encontradas en el .db:", tablas)
```

```python
for nombre_tabla in tablas:
    print(f"\n📄 Contenido de la tabla: {nombre_tabla}")
    try:
        cursor.execute(f"SELECT * FROM {nombre_tabla}")
        registros = cursor.fetchall()
        if registros:
            for fila in registros:
                print(dict(fila))
        else:
            print("   (Sin registros)")
    except Exception as e:
        print(f"   Error al leer tabla {nombre_tabla}: {e}")
```
